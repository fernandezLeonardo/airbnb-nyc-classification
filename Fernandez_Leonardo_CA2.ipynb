## Airbnb NYC Dataset Classification

```python
# Load the dataset and explore it
import pandas as pd

df = pd.read_csv('airbnbData.csv')
print("Number of rows and columns:", df.shape,
      "\nColumns:", df.columns,"\n")
display(df.head(10))

display(df.info())
display(df.describe())
```

```python
# Inspect superhost
display(df['host_is_superhost'].value_counts())
display(df['host_is_superhost'].head(10))
```

```python
# These are all the features, i.e. the columns that are not the target variable (whether host_is_superhost)
display(df.drop(['host_is_superhost'], axis=1).columns)
```

```python
# One hot encode non numerical features
to_encode = [col for col in df.drop(['host_is_superhost'], axis=1).columns if (df[col].dtype == 'str' or df[col].dtype == 'bool')]
print(to_encode,"\n~~~~~~~~~~~~~")
print("Number of unique values in columns:")
print(df[to_encode].nunique())

from sklearn.preprocessing import OneHotEncoder
enc = OneHotEncoder(sparse_output=False)
df_enc = pd.DataFrame(enc.fit_transform(df[to_encode]))
display(df_enc.head(10))
# Reinstate the original column names
df_enc.columns = enc.get_feature_names_out(to_encode)
display(df_enc.head(10))

df = pd.concat([df.drop(to_encode, axis=1).reset_index(drop=True), df_enc.reset_index(drop=True)], axis=1)
display(df.columns[-(df_enc.shape[1]):])
display(df.shape)
```

```python
# Split dataframe into features and target columns
X = df.drop(['host_is_superhost'], axis=1)
y = df[['host_is_superhost']]

print(f"Number of examples: {X.shape[0]}\nNumber of features: {X.shape[1]}")

# Split dataset into training and testing sets (66-33) split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42, stratify=y)
print(f"Training size: {X_train.shape} \nTesting set size: {X_test.shape}")
```

```python
# Implement the decision tree classifier with parameterized function
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
def train_test_DT(X_train, y_train, X_test, y_test, max_depth,leaf=1, crit='entropy'):
    tree = DecisionTreeClassifier(criterion=crit, min_samples_leaf=leaf, max_depth=max_depth)
    tree.fit(X_train, y_train)
    y_pred = tree.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    return accuracy

# Experiment making multiple trees, varying by max depth hyperparameter, and track accuracies to plot
hyperparams = [8, 32]
acc_scores = []
for i in hyperparams:
    acc = train_test_DT(X_train, y_train, X_test, y_test, max_depth=i)
    print(f"Accuracy = {acc}, max_depth = {i}")
    acc_scores.append(acc)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
def visualize_accuracy(hyperparams, acc_scores):
    if len(hyperparams) != len(acc_scores):
        return "There must be one hyperparameter value per accuracy score."
    ax = sns.lineplot(x=hyperparams, y=acc_scores, marker='o')
    ax.set_xlabel('Hyperparameter')
    ax.set_ylabel('Accuracy Score')
    plt.show()

visualize_accuracy(hyperparams, acc_scores)
```

This graph alone does not provide enough information to determine a max_depth value of a high performing model. Although the model seems to decrease in accuracy as the max depth increases, this is only with two data points which vary by a factor of 4. It could be the case that max depth > 8 but < 32 will increase the accuracy. Also, a max depth < 8 might pull the leftmost point higher in accuracy.

```python
# Train various other models
hyperparams = [1, 2, 4, 8, 16, 32]
acc_scores = []
for i in hyperparams:
    acc = train_test_DT(X_train, y_train, X_test, y_test, max_depth=i)
    print(f"Accuracy = {acc}, max_depth = {i}")
    acc_scores.append(acc)
visualize_accuracy(hyperparams, acc_scores)
```

Based on this expanded plot, I would select max_depth = 8 as the hyperparamter for my decision tree, it shows a maximum across the various other values. However, to further explore the possibility of training a higher performing model, I would tune the min_samples_leaf and criterion hyperparameters, which may change the overall model accuracy.

```python
from sklearn.neighbors import KNeighborsClassifier
def train_test_knn(X_train, y_train, X_test, y_test, k):
    knclassifier = KNeighborsClassifier(n_neighbors=k)
    knclassifier.fit(X_train, y_train.to_numpy().ravel())
    y_pred = knclassifier.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    return accuracy

# Now, the hyperparameter to vary is k neighbors in the classifier
hyperparams = [3, 30, 300]
acc_scores = []
for i in hyperparams:
    acc = train_test_knn(X_train, y_train, X_test, y_test, k=i)
    print(f"Accuracy = {acc}, n_neighbors = {i}")
    acc_scores.append(acc)
visualize_accuracy(hyperparams, acc_scores)

# Repeat with more values to visualize
k_range = [i for i in range(1,40,3)]
acc_scores = []
for i in k_range:
    acc = train_test_knn(X_train, y_train, X_test, y_test, k=i)
    print(f"Accuracy = {acc}, n_neighbors = {i}")
    acc_scores.append(acc)
visualize_accuracy(k_range, acc_scores)
```

During testing, the decision tree had a highest accuracy score of ~83% with a max_depth of 8, while the KNN classifier had a high of ~78% with 25 neighbors. Therefore, the decision tree performed better. To find an overall better performing model, min_samples_leaf sizes and criterion can be varied for DTs, as mentioned. For KNN classifiers, the weight function can be changed to give less weight to further points from the input, 'p' (the distance metric) can also be tuned.
